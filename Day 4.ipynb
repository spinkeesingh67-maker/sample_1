```python
class Student:
    def __init__(self, name, roll_no, course, branch):
        self.name = name
        self.roll = roll_no
        self.course = course
        self.branch = branch

    def display(self):
        print("Welcome to Python")
        print(f"Name: {self.name}")
        print(f"Roll no.: {self.roll}")
        print(f"Course: {self.course}")
        print(f"Branch: {self.branch}")
        print("-" * 20) 
        

s1 = Student("Ayukt Yadav", 26, "B.Tech", "Information Technology")
s2 = Student("Rahul Sharma", 30, "B.Tech", "Computer Science")

s1, s2 = s2, s1

s1.display()
s2.display()
```

```python
# Parent Class (Base Class)

class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_person(self):
        print("Name :", self.name)
        print("Age  :", self.age)


# Child Class (Derived Class)

class Student(Person):

    def __init__(self, name, age, roll_no, marks):
        # Call Parent Constructor
        super().__init__(name, age)

        # Child Class Variables
        self.roll_no = roll_no
        self.marks = marks

    def show_student(self):
        print("Roll Number :", self.roll_no)
        print("Marks       :", self.marks)


# Create Object

student1 = Student("Aryaan", 20, 101, 92)


# Display Data

print("Student Information")
print("--------------------")

student1.show_person()      # Parent Class Method
student1.show_student()     # Child Class Method
```
